### 目标检测（Detection）、图像分割（Segmentation）、可视化（Visualization）
#### 语义分割（Semantic Segmentation）
语义分割本质上是为图片的每个像素进行分类，但如果直接使用CNN，则会导致计算量过大；一个解决方案是引入全卷积神经网络（Fully Convolutional），它首先通过卷积层进行下采样，得到分辨率较低但通道数很多的中间层，然后通过上采样恢复到原来的分辨率（类别数*分辨率），即对每个像素，有一个它对每个类别的打分

#### 目标检测（Detection）和实例分割（Instance Segmentation）
* 目标检测的一种方法是R-CNN，R指代Region，即先指定一些Bounding Box，然后在这些Region中做卷积神经网络，看看框出来的东西是什么。但这样做的一个问题是计算时间过长，一个解决方法是先在整个图像上做卷积神经网络，得到特征图，然后在特征图上做上述操作
* YOLO 把图片分成一格一格的。每个格子都会预测若干个框。训练时，如果某个物体的中心点落在某个格子里，就由这个格子负责预测该物体；在这个格子的 B 个预测框中，选和真实框 IoU 最大的那个框来学习这个目标
* Object Detection with Transformer：DETR 先用 CNN 提取图像特征，再用 Transformer encoder-decoder 处理全局关系，通过一组 object queries 直接输出固定数量的目标预测；训练时用 bipartite matching 把预测框和真实框一一匹配，匹配上的预测学习类别和框位置，没匹配上的预测学习 no object

#### 可视化（Visualization）和理解（Understanding）
可视化和理解主要是为了弄明白各种神经网络的作用原理，例如，显著图（Saliency Maps）就是通过计算分类得分关于每个像素的梯度，找到那些最影响分类结果的像素绘制而成的；如果 CNN 最后一层卷积特征经过 Global Average Pooling 再接全连接分类层，那么我们可以用分类层中某个类别的权重去加权最后一层特征图，从而得到这个类别的激活热力图，也就是 CAM，用来观察模型判断该类别时主要看图像的哪些区域
