### 线性分类器进行图像分类
#### 核心任务：图像分类
* 任务解释：已知一些标签labels（猫狗），尝试给每个image分配一个合适的标签
* 计算机定义图像为数据矩阵（张量）：每个像素取值0-255，共有三个通道（红绿蓝）；例如对于一张分辨率为800\*600的图片，其张量大小为800\*600\*3
#### Data-Driven Approach
##### 数据集：CIFAR10
##### 一般步骤
1. Collect a dataset of images and labels
2. Use Machine Learning algorithms to train a classifier（train函数）
3. Evaluate the classifier on new images（predict函数）
##### Nearest Neighbor Classifier
* 对于近邻算法，train函数用于存储所有数据和标签，predict函数用于寻找训练集中最接近测试集的图片并将标签赋予给这个测试集图片
* 用距离函数衡量两张images之间的距离：$L_1$距离（矩阵相减，然后所有元素相加）；$L_2$距离（欧氏距离）。相比于$L_2$距离，$L_1$距离保留特征位置的意义，例如如果旋转特征轴，$L_1$距离会发生变化
* 算法复杂度：train O(1)；predict O(N)，这说明近邻算法很复杂，计算速度很慢
* 可视化http://vision.stanford.edu/teaching/cs231n-demos/knn/
* 超参数的确定：（1）确定k值；（2）选择哪种距离函数。<br>
方法一：基于训练集确定（不好，因为永远在训练集上表现完美）<br>
方法二：基于测试集确定（不好，无法适应测试集以外的数据）<br>
方法三：划分训练集和验证集，基于验证集确定（更好的方式）<br>
方法四：k折交叉验证（在深度学习中并不常用，成本太高）<br>
方法五：直觉！

* 实际上就是对线性计算的结果先做指数运算，再归一化，损失函数为交叉熵


```python
# Nearest Neighbor Classifier
import numpy as np

class NearestNeighbor:
    def __init__(self):
        pass

    def train(self, X, y):
        # X是一个N x D的矩阵，N是训练样本的数量，D是每个样本的维度
        # y是一个长度为N的向量，包含每个训练样本的标签
        # 存储数据
        self.Xtr = X
        self.ytr = y

    def predict(self, X):
        # X是一个M x D的矩阵，M是测试样本的数量，D是每个样本的维度
        num_test = X.shape[0]
        # 确保预测结果格式符合训练标签的格式
        Ypred = np.zeros(num_test, dtype = self.ytr.dtype)

        for i in range(num_test):
            """
            计算第i个测试样本与所有训练样本之间的L_1距离，
            得到的distances是一个长度为N的向量，
            包含第i个测试样本与每个训练样本之间的距离
            """
            distances = np.sum(np.abs(self.Xtr - X[i,:]), axis = 1)
            min_index = np.argmin(distances)# 这是训练集中的索引
            Ypred[i] = self.ytr[min_index]

        return Ypred
```
