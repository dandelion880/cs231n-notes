### 三维视觉
直接显式表示三维物体的方法包括点图，网状图，以及参数方法，例如函数映射；隐式表示包括使用体素（voxels，类比图片的像素）或构造函数 $f$，用 $f = 0$ 表示一个三维曲面，它的好处是很容易判断某个点是在物体内还是物体外，但缺点是难以采样，因为需要解方程 $f = 0$ 。<br>
三维视觉深度学习的任务包括：生成和分类，将3D和2D结合（例如3D和2D相互投影、利用2D预训练模型帮助3D任务），将3D与其它模态结合，如文字等
#### 3D深度学习模型
* 受2D图片CNN的启发，一个自然的思路是利用3D样本的voxel图，将3D卷积应用在上面，从而进行分类和生成等任务
* PointNet 是第一个非常经典的、直接吃点云的深度学习模型。它用共享 MLP 处理每个点，再用 max pooling 聚合成全局特征，从而解决点云无序、不规则的问题
* AtlasNet 是一种把 3D 形状表示成多个可学习的二维曲面 patch 的生成模型。它用神经网络学习从二维参数域到三维空间的映射，从而生成或重建物体表面
* NeRF 用神经网络 $F_\Theta(x,y,z,\theta,\phi)$ 表示三维场景中每个点的颜色和密度；训练时从 2D 图片发射 ray，沿 ray 查询网络并渲染出像素颜色，再和真实图片比较，从而学出一个可以从新视角渲染的 3D 场景
