### 生成模型（Generative Models）（一）
| 模型类型   | 学什么           | 典型问题            |
| ------ | ------------- | --------------- |
| 判别模型   | $p(y \mid x)$ | 给图片分类：这是猫还是狗？   |
| 生成模型   | $p(x)$        | 学真实图片分布，然后生成新图片 |
| 条件生成模型 | $p(x \mid y)$ | 给定“猫”，生成一张猫图    |

生成模型的意义：当一个输入条件可以对应很多合理输出时，生成模型可以学习“这些输出的概率分布”，而不是强行预测唯一答案

#### Autoregressive Models
基于极大似然估计设计损失函数：给定真实图片数据集（i.i.d）$x^{(1)},x^{(2)},\dots,x^{(N)}$，我们的目标是训练神经网络$f(x,W)$使得每个$p(x^{(i)})$尽可能大，因此
$$W^* = \argmax_{W}\Pi_i p(x^{(i)})$$
从而损失函数可以设计为
$$\mathbf{L} = -\sum_{i} \log f(x^{(i)},W)$$
又设输入的每个样本$x$是一个序列$(x_1,x_2,\dots,x_T)$，则可根据链式法则将$P(x)$写作
$$p(x) = \Pi_{t=1}^T p(x_t|x_1,\dots,x_{t-1})$$
这样就使得神经网络更容易构建（拆分成小部分）

#### Variational Autoencoders (VAEs)
Autoencoder 通过 encoder 将输入 $x$ 编码为表示 $z$，再通过 decoder 将 $z$ 解码为重建结果 $\hat{x}$，通过衡量 $x$ 和 $\hat{x}$ 的相似程度训练 encoder 和 decoder。如果想把它当生成模型使用，就希望丢掉 encoder，只用 decoder 从新的 z 生成样本。但普通 Autoencoder 的问题是：它没有规定 $z$ 的分布，因此我们不知道如何采样合理的 $z$。VAE 的核心动机就是约束潜变量 $z$ 接近某个简单分布，例如标准正态分布，从而训练后可以从该分布随机采样 $z$，再通过 decoder 生成新数据

VAE的训练过程可以简述为：给定数据$x$，首先经过encoder得到一个$z$的分布 $q_\phi(z|x)$，力求这个分布尽可能接近标准正态分布（这正是上图中损失函数第二项衡量的事），然后从这个分布中采样$z$，经过decoder得到一个$x$的分布 $p(x|z)$，将均值作为输出，直观上每个像素填入均值就得到了生成的图片，力求这个图片与原始输入$x$接近（这正是上图中损失函数第一项衡量的事）。以此来同时训练encoder和decoder得到VAE
