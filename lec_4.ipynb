### 神经网络与反向传播
#### 神经网络
* 引入多层神经网络的原因：对于线性分类器不可分类的数据，先通过某种非线性变换将它们变成线性分类器可分类的
* 激活函数（Activation Function）<br>
作用：如果没有激活函数，那么多层神经网络退化为线性分类器<br>

* 上面的激活函数中，tanh和sigmoid可能会导致梯度消失，最常用的是ReLU
* 更多层的神经网络往往意味着更复杂的模型，但可能会导致过拟合，这时可以提高正则化强度
#### 反向传播（Backpropagation）
定义：沿着计算图应用链式法则计算所有输入/参数/中间量的梯度；关于各参数的梯度维数应与原始输入相同<br>
前向：计算一个operation的结果并存储用于计算梯度的中间量<br>
反向：即反向传播

```python
# 一个使用sigmoid作为激活函数的三层神经网络示例（前向传播）
import numpy as np

f = lambda x:1.0/(1.0+np.exp(-x))
x = np.random.randn(3, 1)
W1 = np.random.randn(4, 3)
b1 = np.random.randn(4, 1)
W2 = np.random.randn(4, 4)
b2 = np.random.randn(4, 1)
W3 = np.random.randn(1, 4)
b3 = np.random.randn(1, 1)
h1 = f(np.dot(W1, x) + b1)
h2 = f(np.dot(W2, h1) + b2)
out = np.dot(W3, h2) + b3

print(out)
```

```python
# 一个使用sigmoid作为激活函数的两层神经网络训练过程示例
import torch

N, D_in, H, D_out = 64, 1000,100, 10 # （批次大小，输入维度，隐藏层维度，输出维度）
x, y = torch.randn(N, D_in), torch.randn(N, D_out)
w1, w2 = 0.01 * torch.randn(D_in, H), 0.01 * torch.randn(H, D_out)
learning_rate = 1e-3

for i in range(1000):
    # 前向传播
    h = 1/(1+torch.exp(-x.mm(w1))) # sigmoid函数
    y_pred = h.mm(w2)
    loss = (y_pred - y).pow(2).mean().item()
    if i % 100 == 99:
        print(i, loss)
    
    # 反向传播
    grad_y_pred = 2.0 * (y_pred - y)/N
    grad_w2 = h.t().mm(grad_y_pred)
    grad_h = grad_y_pred.mm(w2.t())
    grad_w1 = x.t().mm(grad_h * h * (1 - h)) # sigmoid函数的导数

    # 更新权重
    w1 -= learning_rate * grad_w1
    w2 -= learning_rate * grad_w2
```
