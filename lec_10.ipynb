### 视频理解（Video Understanding）
#### 基于CNN的方法
一种基于图片分类的方法是对于一段视频，随机采样几帧，对每一帧做2D CNN得到特征图（T\*D\*H'\*W'），然后将它们展平并拼接在一起，最后用一个MLP得到每一类的分数（Late Fusion）。但这种方式会导致最后拼接到一起的向量很大，因此参数过多，效率降低，一个解决方案是用pooling取代展平拼接。另一种方案是首先将输入（T\*3\*H\*W）reshape成（3T\*H\*W），然后进行2D CNN（Early Fusion）<br>
2D Early Fusion把多帧视频一开始就压成类似多通道图片来处理，简单，但时间维度被固定住了，不具备时间平移不变性，不同帧里的同一个动作可能需要不同的卷积核重复学习；3D CNN卷积核在时间和空间上一起滑动，可以更自然地学习动作、运动、变化等视频特征。3D CNN也可以搭建和VGG类似的结构

#### 更关注Motion的方法：光流法（Optical Flow）
视频中对象的外貌和动作可以分开来看：外貌流只关心空间变化，动作流只关心时间变化

#### RNN和自注意力机制的应用
我们也可以通过RNN来处理视频，例如进行视频分类时，只需在最后输出一个类别，但RNN无法并行运算
