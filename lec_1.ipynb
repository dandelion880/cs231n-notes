### 课程总览
#### Deep Learning Basics
* 图片分类：线性分类器（找到超平面分类）；回归和优化；神经网络（非线性分类问题）
#### Perceiving and Understanding the Visual World
* 语义分割（Semantic Segmentation）：为每个像素标签
* 目标检测：不仅分类，还要确定物体的位置
* 实例分割（Instance Segmentation）：结合语义分割和目标检测，每个实例对象都有自己的掩码（mask）
* 视频分类；多模态理解（Multimodal Video Understanding）；注意力机制
* 模型：CNN；RNN；Transformer
#### Generative and Interactive Visual Intelligence
* 自监督学习（Self-supervised Learning）：处理不带标签的数据
* 生成模型（Generative Modeling）：风格迁移
* 扩散模型（Diffusion Model）：去噪
* 视觉语言模型（Vision Language Models）：文生图/图生文
* 3D Vision;具身智能
#### Humen-Centered Applications and Implications
