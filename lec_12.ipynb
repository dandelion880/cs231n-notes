### 自监督学习
大规模训练往往需要大量人工标注的数据，因此人们希望绕过这个过程，这就是自监督学习（Self-Supervised Learning）：先在无标签数据上设计一个“自动有答案”的预训练任务（Pretext Task），让模型学到通用表示；然后把学到的 encoder 迁移到有标签的下游任务中。例如：有一组无标签图片，我们人为构造一个“有答案”的任务，比如把图片旋转 0°，90°，180°，270°，让模型预测旋转角度。虽然最终任务不是识别角度，但模型为了完成这个任务，必须学会提取图像中的形状、方向、物体结构等特征。之后我们把这个已经训练好的 encoder 迁移到下游任务，比如图像分类、检测、分割。<br>
衡量一个自监督学习模型的标准：在Pretext Task上的表现；泛化能力；计算效率；迁移学习和下游任务的表现（最重要）

#### Pretext Task
* 人为将无标签图片旋转特定角度，训练模型预测图片旋转角度：我们假设模型为了学到旋转角度，必须学会识别整个物体
* 将图片分为若干patches，训练模型预测相关patch的相对位置，或预测打乱的patches的排序
* 人为mask图片的一部分，训练模型预测缺失像素（inpainting）
* 图片着色：给定图片的$L$通道，训练模型预测$ab$通道；一个推广是视频着色，因为给定参考帧后，模型为了给后续帧着色，需要学会追踪某个区域或物体
* Masked Auto Encoders（MAE）：MAE 是一种基于 ViT 的自监督预训练方法。它先把图片切成 patch，然后随机遮住约 75% 的 patch，只把剩下的 25% 输入 encoder。encoder 提取可见部分的特征，decoder 尝试重建被遮住的图像内容。由于遮挡比例很高，模型必须学习图像的整体结构和语义，因此训练出的 encoder 可以迁移到下游视觉任务中

#### Constrastive Representation Learning
通过构造“相似样本”和“不相似样本”，训练 encoder 把相似对象映射到相近的特征向量，把不相似对象映射到较远的特征向量，从而学到可迁移的表示
* SimCLR：它对每张无标签图片生成两个增强版本，把它们当作正样本对，把其他图片当作负样本，通过对比损失训练 encoder，使同一图片的不同视角特征接近、不同图片的特征远离，从而学到可迁移的视觉表示
