### 基于CNN的图像分类（Convolutional Networks）
有时人们并不直接把原始数据张量（例如图片的像素值 Raw Pixels 展平）作为神经网络的输入，而是先进行一定的预处理。曾经的一类方法是特征提取（Feature Extraction），例如提取一张图片的颜色分布直方图，但这样会破坏图片的空间结构；或者忽略颜色信息，只看空间特征；也可以把提取出来的不同特征组合在一起，形成一个张量作为输入。但比起人为设计特征提取算法，近年来更有效的方法是仍然将原始像素值作为输入，让机器通过优化自动进行特征提取（End to End），即仅设计一个神经网络框架，但里面的参数由机器学习完成<br>
* 直接展平像素作为输入的问题：图像的空间结构被破坏了
#### 卷积神经网络CNN
* 卷积层：卷积层的作用就可以理解为机器自动提取特征，这一过程保留一定的空间结构<br>
卷积核（Filter）理解：卷积核的深度必须等于输入通道数；一个卷积核输出一个通道，多个卷积核输出多个通道。例如，对一个32*32*3的图片，使用64个5*5*3的卷积核，就是每个卷积核有3个5*5的二维矩阵，分别与原始输入的三个通道作积求和，最后将三个结果加起来得到一个二维矩阵，64个卷积核意味着64个输出通道，即64个二维矩阵。类似全连接神经网络，每个卷积核有一个bias

* 注意，每个卷积层需要使用激活函数，否则模型将是线性的；训练开始时，每个卷积核是随机初始化的，这也避免不同卷积核提取相同特征
* 矩阵大小的计算：设输入为$W$，卷积核大小为$K$，那么输出为$W-K+1$。由此可见卷积运算可能会使矩阵（Feature map）变小，为了保持其保持不变，一种解决方法是引入填充（Padding），即在周围填$0$，设填充为$P$，则输出为$W-K+1+2P$，因此通常卷积核大小$K$为奇数，$P=\frac{K-1}{2}$
* 感受野（Receptive Fields）：感受野是指更深层的卷积层中的某个元素受原始输入中哪些像素的影响，一般来说越深的卷积层感受野越大。我们通常希望最终的感受野包含整个原始输入，为了通过较少层达到这个效果，一个方法是引入步长（Stride），设步长为$S$，则输出为$\frac{W-K+2P}{S+1}$

#### 池化层（Pooling Layers）
* 池化的作用和步长一样，是为了Downsample，即减小CNN的复杂度，加速处理过程
* Max Pooling：例如若想将维数减少为原来的一半，则使用一个2\*2的卷积核，以2为步长作用在输入上，作用方式是取输入上每个2\*2方阵中的最大值；这一过程没有需要学习的参数
* 卷积和池化的性质：Translation Equivariance，即先对图像做平移，再做卷积或池化，应当与先卷积或池化，再平移得到的结果相同；这说明卷积神经网络在进行分类时不受物体在画面中绝对位置的影响
